```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

file_path = r"C:\Users\Suresh seervi\Desktop\t3\WorldPopulation.csv"

df = pd.read_csv(file_path)

print("First 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

df = df.sort_values("Year")

df["Population_Billions"] = df["Population"] / 1_000_000_000
df["Urban_Billions"] = df["Urban"] / 1_000_000_000

print("\nPopulation Growth:")
print(df[["Year", "Population", "Population_Billions", "ChangePerc"]].to_string(index=False))

max_population = df.loc[df["Population"].idxmax()]

print("\nYear with Highest Population:")
print(max_population["Year"])

print("\nHighest Population:")
print(f"{max_population['Population_Billions']:.2f} Billion")

min_population = df.loc[df["Population"].idxmin()]

print("\nYear with Lowest Population:")
print(min_population["Year"])

print("\nLowest Population:")
print(f"{min_population['Population_Billions']:.2f} Billion")

print("\nAverage Population:")
print(f"{df['Population_Billions'].mean():.2f} Billion")

print("\nAverage Population Growth Rate:")
print(f"{df['ChangePerc'].mean():.2f}%")

print("\nYear with Highest Growth Rate:")
highest_growth = df.loc[df["ChangePerc"].idxmax()]
print(highest_growth["Year"])
print(f"Growth Rate: {highest_growth['ChangePerc']:.2f}%")

plt.figure(figsize=(10, 6))
plt.plot(df["Year"], df["Population_Billions"], marker="o")
plt.title("World Population Growth Over the Years")
plt.xlabel("Year")
plt.ylabel("Population (Billions)")
plt.grid(True)
plt.tight_layout()

plt.savefig('World Population Growth Over the Years.png')
plt.show()

plt.figure(figsize=(10, 6))
plt.bar(df["Year"], df["Population_Billions"])
plt.title("World Population by Year")
plt.xlabel("Year")
plt.ylabel("Population (Billions)")
plt.tight_layout()

plt.savefig('World Population by Year.png')
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(df["Year"], df["ChangePerc"], marker="o")
plt.title("World Population Growth Rate")
plt.xlabel("Year")
plt.ylabel("Annual Growth Rate (%)")
plt.grid(True)
plt.tight_layout()

plt.savefig('World Population Growth Rate.png')
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(df["Year"], df["NetChange"], marker="o")
plt.title("Annual Population Net Change")
plt.xlabel("Year")
plt.ylabel("Population Increase")
plt.grid(True)
plt.tight_layout()

plt.savefig('Annual Population Net Change.png')
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(df["Year"], df["Density"], marker="o")
plt.title("Population Density Over the Years")
plt.xlabel("Year")
plt.ylabel("Population Density")
plt.grid(True)
plt.tight_layout()

plt.savefig('Population Density Over the Years.png')
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(df["Year"], df["Urban_Billions"], marker="o")
plt.title("Urban Population Growth")
plt.xlabel("Year")
plt.ylabel("Urban Population (Billions)")
plt.grid(True)
plt.tight_layout()

plt.savefig('Urban Population Growth.png')
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(df["Year"], df["UrbanPerc"], marker="o")
plt.title("Urban Population Percentage")
plt.xlabel("Year")
plt.ylabel("Urban Population (%)")
plt.grid(True)
plt.tight_layout()


plt.savefig('Urban Population Percentage.png')
plt.show()

plt.figure(figsize=(10, 6))
plt.hist(df["Population_Billions"], bins=10, edgecolor="black")
plt.title("Distribution of World Population")
plt.xlabel("Population (Billions)")
plt.ylabel("Frequency")
plt.tight_layout()

plt.savefig('Distribution of World Population.png')
plt.show()

print("\nFinal Insights:")
print("1. World population has increased continuously over the given period.")
print("2. Population increased from around", round(df["Population_Billions"].min(), 2), "billion to", round(df["Population_Billions"].max(), 2), "billion.")
print("3. The annual population growth rate has generally decreased over time.")
print("4. Urban population has increased significantly over the years.")
print("5. The percentage of people living in urban areas has also increased.")
print("6. Population density has increased as the world population has grown.")
```
